```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[5,6,7,8,9]

plt.plot(x,y)
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title("Basic line plot")
plt.show()
```

```python
x=[1,2,3,4,5]
y=[5,6,7,8,9]

#Customized line plot

plt.plot(x,y,color="Black",linestyle='-.',marker='o')   #marker=cordinates of x and y
```

```python
plt.plot(x,y,color="Black",linestyle='-.',marker='o',linewidth=3,markersize=9) 

plt.grid(True)
```

```python
X=[2,3,4,5]
Y=[7,9,8,7]
Z=[8,9,1,2]


plt.subplot(2,2,1)
plt.plot(X,Y,color='Red',marker='o',linestyle='-')
plt.title("Plot 1")


plt.subplot(2,2,2)
plt.plot(X,Z,color="Black",marker='*',linestyle='--')
plt.title("Plot 2")

plt.subplot(2,2,3)
plt.plot(Y,Z,color="Blue",marker='o',linestyle='-.')
plt.title("Plot 3")

plt.subplot(2,2,4)
plt.plot(Y,X,color="Green",marker='*',linestyle='--')
plt.title("Plot 4")
```

BAR GRAPH

```python
Category=['OC','BC','ST','EWS']
Population=[10.5,20.9,30.77,40.9080]

plt.bar(Category,Population,color='purple')
plt.xlabel("Category")
plt.ylabel("Population")
plt.title("Bar plot")
```

HISTOGRAM


```python
x=[1,2,2,3,3,3,4,4,4,4]
plt.hist(x,bins=7,color="orange",edgecolor="black")
```

SCATTER PLOT


```python
x=[1,2,3,4,5]
y=[2,3,4,5,6]

plt.scatter(x,y,color="blue",marker='*')
```

PIE CHATS

```python
labels=['A','B','C','D','E']
sizes=[20,30,40,50,60]
colors=['gold','yellowgreen','red','blue','orange']
explode=[0.2,0,0,0,0]

plt.pie(sizes,explode=explode,labels=labels,colors=colors,autopct="%1.1f%%",shadow=True)
```
